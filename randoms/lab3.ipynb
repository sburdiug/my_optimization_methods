```python
import numpy as np
import matplotlib.pyplot as plt

# Красивый вывод (4 знака)
np.set_printoptions(precision=4, suppress=True)

# -------------------------------
# 1. Дискретний розподіл (варіант 4)
# -------------------------------
x = np.array([-3, -2, -1, 0, 1, 2, 3])
p = np.array([0.2, 0.15, 0.1, 0.1, 0.3, 0.05, 0.1])

cdf = np.cumsum(p)

print("x =", x)
print("p =", p)
print("F(x) =", cdf)

# Генерація
y = np.random.rand(100)
sample = x[np.searchsorted(cdf, y)]

print("\nВибірка (100):")
print(sample)

# -------------------------------
# 2. Частоти
# -------------------------------
n = 100
empirical_counts = np.array([np.sum(sample == xi) for xi in x])
theoretical_counts = n * p

print("\nЕмпіричні частоти:", empirical_counts)
print("Теоретичні частоти:", np.round(theoretical_counts, 4))

# Графік
plt.figure(figsize=(8, 5))
plt.bar(x, empirical_counts, width=0.6, alpha=0.7, edgecolor='black')
plt.plot(x, theoretical_counts, 'o-', linewidth=2)
plt.title('Дискретна випадкова величина')
plt.grid(True, alpha=0.3)
plt.show()

# -------------------------------
# 3. Експоненційний розподіл
# -------------------------------
team_number = 4   # <-- поменяй если нужно
lam = team_number / 10

y_exp = np.random.rand(100)
exp_sample = -np.log(1 - y_exp) / lam

print("\nλ =", lam)
print("Експоненційна вибірка:")
print(np.round(exp_sample, 4))

# Графік
plt.figure(figsize=(8, 5))
plt.hist(exp_sample, bins=12, density=True, alpha=0.7, edgecolor='black')

x_vals = np.linspace(0, max(exp_sample), 300)
f_vals = lam * np.exp(-lam * x_vals)

plt.plot(x_vals, f_vals, linewidth=2)
plt.title('Експоненційний розподіл')
plt.grid(True, alpha=0.3)
plt.show()

# -------------------------------
# 4. Потік подій
# -------------------------------
times = exp_sample[:50]
events = np.cumsum(times)

print("\nІнтервали часу:")
print(np.round(times, 4))

print("\nМоменти подій:")
print(np.round(events, 4))

plt.figure(figsize=(10, 2.5))
plt.eventplot(events)
plt.title('Потік подій')
plt.grid(True, alpha=0.3)
plt.show()
```
