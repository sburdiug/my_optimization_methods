```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("лаб1.csv", header=None, sep=None, engine="python")

df = df.replace({',': '.'}, regex=True)

# Перетворимо все, що можна, в числа (що не можна -> NaN)
df = df.apply(pd.to_numeric, errors="coerce")

# Викидаємо повністю порожні колонки
df = df.dropna(axis=1, how="all")

print("Форма після очищення:", df.shape)

# Якщо колонок понад 10 — зазвичай перша колонка це індекс/номер реалізації.
# Беремо останні 10 колонок як перерізи.
if df.shape[1] > 10:
    df = df.iloc[:, -10:]

# Якщо менш як 10 — файл, ймовірно, зчитався некоректно (роздільник/формат), потрібна інша стратегія.
if df.shape[1] != 10:
    raise ValueError(f"Після обробки вийшло {df.shape[1]} колонок, а треба 10. "
                     f"Перевір роздільник (',' чи ';') і формат чисел.")

data = df.to_numpy()  # 100×10 очікуємо

# ---------- 1) Вибір 30 реалізацій (варіант 4) ----------
variant4 = [2,6,8,14,15,18,22,24,28,31,35,39,42,45,46,
            51,55,60,62,64,71,73,78,81,84,88,91,96,98,99]
idx = [i - 1 for i in variant4]
selected = data[idx, :]  # 30×10

n = selected.shape[0]
```

```python
plt.style.use("default")

def sturges_k(n: int) -> int:
    return int(np.ceil(1 + np.log2(n)))

def my_plot(ax, values: np.ndarray, edges: np.ndarray, title: str):
    counts, _ = np.histogram(values, bins=edges)
    widths = np.diff(edges)          # Δ (усі однакові)
    heights = counts / widths        # n_i / Δ

    ax.bar(
        edges[:-1],
        heights,
        width=widths,
        align="edge",
        edgecolor="black",
        linewidth=1.0
    )

    ax.set_xticks(edges)
    ax.set_xticklabels([f"{x:.2f}" for x in edges], rotation=45, ha="right")
    #ax.set_xlabel("x (межі інтервалів)")
    ax.set_ylabel("h = nᵢ / Δ")
    ax.set_title(title)
    ax.grid(True, axis="y", alpha=0.3)

k = sturges_k(n)             # 6

# Спільні межі для всіх перерізів
global_min = float(np.min(selected))
global_max = float(np.max(selected))
edges_global = np.linspace(global_min, global_max, k + 1)

print("k =", k)
print(f"global_min = {global_min:.4f}, global_max = {global_max:.4f}")
delta = edges_global[1] - edges_global[0]
print(f"Δ = {delta:.4f}, sum(nᵢ) = {n}")

fig, axes = plt.subplots(2, 5, figsize=(22, 8), constrained_layout=True, facecolor="white")
fig.suptitle(f"Спільні межі, k={k}", fontsize=14)
axes = axes.ravel()
for ax in axes:
    ax.set_facecolor("white")
for j in range(10):
    my_plot(axes[j], selected[:, j], edges_global, f"Переріз {j+1}")

plt.show()
```
